```python
from your_job_offer.services.mail_checker.methods import get_email_messages
from your_job_offer.use_cases.tracking.internal import *
from your_job_offer.utils.mock import get_user
    
def print_with_line_breaks(text, line_length=100):
    """Выводит длинную строку с переносами на заданную длину.
    
    :param text: Исходный текст для вывода.
    :param line_length: Максимальная длина строки.
    """
    while len(text) > line_length:
        # Найдем место для разбиения строки, чтобы не разрывать слова
        break_index = text.rfind(' ', 0, line_length)
        if break_index == -1:  # Нет пробела, берем отрезок от начала
            break_index = line_length
        
        print(text[:break_index])  # Выводим часть строки
        text = text[break_index:].lstrip()  # Убираем пробелы и продолжаем

    # Выводим оставшуюся часть текста
    print(text)
    print()

def print_messages(messages):
    for message in messages:
        print_with_line_breaks(message.body)

user = get_user()
messages = get_email_messages(user)
print_messages(messages)
filtered = filter_from_spam(messages)
cleaned = clean_messages(filtered)
```

```python
from your_job_offer.use_cases.tracking.email_parser import parser

def to_str(msg):
    if msg is None:
        return None
    return f"{msg.vacancy_key.job}, {msg.vacancy_key.employer}, {msg.stage.stage_type}"

ans = parser.parse(cleaned)
for clean, parsed in zip(cleaned, ans):
    print("Сообщение")
    print_with_line_breaks(clean.body)
    print(f"Результат: {to_str(parsed)}")
    print("------------------------\n")
```
